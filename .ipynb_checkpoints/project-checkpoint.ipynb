```python
# LOAD RELEVANT LIBRARIES (may need more?)
import numpy as np
import time
from matplotlib.colors import ListedColormap, BoundaryNorm
from sklearn.neural_network import MLPClassifier
from sklearn.inspection import DecisionBoundaryDisplay
from sklearn.preprocessing import LabelEncoder, OrdinalEncoder, OneHotEncoder
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import f1_score, classification_report, r2_score
import tensorflow as tf
import matplotlib.pyplot as plt
seed=0; 
np.random.seed(seed)  # Random number seed
```

```python
# DATASET INFO
FILENAME = 'dataset copy.csv'
NUM_EPOCHS = 5
```

```python
# READ IN DATA
X = np.loadtxt(FILENAME, delimiter=',', skiprows=1, usecols=range(5,14), dtype=float)
y = np.loadtxt(FILENAME, delimiter=',', skiprows=1, usecols=15, dtype=str, encoding='utf-8')

# SPLIT INTO TRAINING/VALIDATION/TESTING (70/15/15)
X_train, X_other, y_train, y_other = train_test_split(X, y, train_size=0.7, random_state=seed)
X_val, X_test, y_val, y_test = train_test_split(X_other, y_other, test_size=0.5, random_state=seed)

# FEATURE SCALING
from sklearn.compose import ColumnTransformer
#need this ^ to skip over columns?

scaler = StandardScaler().fit(X_train)

# popularity (column 0) and tempo (column 7) need to be scaled
# Scale only columns 0 and 7; pass other columns through unchanged
ct = ColumnTransformer(
    transformers=[('scaler', StandardScaler(), [0, 7])],
    remainder='passthrough'  # Keeps untouched columns instead of dropping them
)

X_train = ct.fit_transform(X_train)
X_val = ct.fit_transform(X_val)
X_test = ct.fit_transform(X_test)


USE_ONE_HOT_ENCODING = False  # False=OrdinalEncoding, True=OneHotEncoding
encoder = OrdinalEncoder()
if (USE_ONE_HOT_ENCODING): encoder = OneHotEncoder()
encoder.fit(y_train.reshape(-1, 1))  # Fit to training data
y_train = encoder.transform(y_train.reshape(-1, 1))  # Encode training data
y_val = encoder.transform(y_val.reshape(-1, 1))  # Encode validation data
y_test = encoder.transform(y_test.reshape(-1, 1))  # Encode testing data

#print('Genre values for first 5 songs after encoding\n', y_train[:5])  # Print first 5 songs
#print('Num features in encoded data:', y_train.shape[1])
#print(encoder.categories_) 
#There are 114 genre values
```
